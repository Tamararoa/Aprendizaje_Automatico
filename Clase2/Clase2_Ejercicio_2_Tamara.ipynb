Ejercicio 2 - 1
Cargar el conjunto de datos en una estructura adecuada para su manipulación.

```python
import pandas as pd
automobile = pd.read_csv ("/content/Automobile.csv", encoding="ISO-8859-1")
automobile.head ()
```

Ejercicio 2 - 2:Crear un gráfico de dispersión para visualizar las relaciones entre los pares de características numéricas `mpg` y `weight`.

```python
import matplotlib.pyplot as plt


plt.figure(figsize=(8, 5))
plt.scatter(automobile["weight"], automobile["mpg"], alpha=0.7, color="blue")

plt.title("Relación entre Weight y MPG")
plt.xlabel("Weight (Peso)")
plt.ylabel("MPG (Millas por Galón)")
plt.grid(True)
```

Ejercicio 2 - 3:Generar un histograma del consumo de combustible (`mpg`) para analizar su distribución.

```python
import matplotlib.pyplot as plt


plt.figure(figsize=(8, 5))
plt.hist(
    automobile["mpg"], bins=15, color="skyblue", edgecolor="black", alpha=0.7
)


plt.title("Distribución del Consumo de Combustible (MPG)")
plt.xlabel("MPG (Millas por Galón)")
plt.ylabel("Frecuencia")
plt.grid(True, linestyle="--", alpha=0.5)


plt.show()
```

Ejercicio 2 - 4:Graficar un diagrama de caja (boxplot) de las características `cylinders` y `origin` para identificar posibles valores atípicos.


```python
import matplotlib.pyplot as plt
import seaborn as sns

# Configurar el tamaño de la figura (1 fila, 2 columnas)
plt.figure(figsize=(12, 5))

# 1. Boxplot de Cylinders vs MPG
plt.subplot(1, 2, 1)
sns.boxplot(data=automobile, x="cylinders", y="mpg", palette="Set2")
plt.title("Valores atípicos: MPG por Cylinders")
plt.xlabel("Cilindros (cylinders)")
plt.ylabel("Millas por galón (mpg)")

# 2. Boxplot de Origin vs MPG
plt.subplot(1, 2, 2)
sns.boxplot(data=automobile, x="origin", y="mpg", palette="Set3")
plt.title("Valores atípicos: MPG por Origin")
plt.xlabel("Origen (origin)")
plt.ylabel("Millas por galón (mpg)")

plt.tight_layout()
plt.show()
```

Ejercicio 2-5:Realizar un gráfico de barras que muestre la distribución de automóviles según su año de modelo (`model_year`) y origen (`origin`).

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Configurar el tamaño de la figura
plt.figure(figsize=(10, 6))

# Crear el gráfico de barras agrupadas
sns.countplot(data=automobile, x="model_year", hue="origin", palette="Set2")

# Personalizar el gráfico
plt.title("Distribución de Automóviles por Año de Modelo y Origen")
plt.xlabel("Año del Modelo (model_year)")
plt.ylabel("Cantidad de Automóviles")
plt.legend(title="Origen")
plt.grid(axis="y", linestyle="--", alpha=0.5)

# Mostrar el gráfico
plt.show()
```
